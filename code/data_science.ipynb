```python
import os
from PIL import Image
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from scipy import stats
```

```python
#Xử lí ảnh qq-plot
IMAGE_DIR = r"C:\Users\ACER\Documents\BTL_XSTK\XSTK_GPU\bin"  
OUTPUT_DIR = "./images"


os.makedirs(OUTPUT_DIR, exist_ok=True)

VAL_SCALE = 0.7

PADDING_BETWEEN = 10 

val_files = [f for f in os.listdir(IMAGE_DIR) if f.endswith("_val.png")]

for val_filename in val_files:
    base_filename = val_filename.replace("_val.png", ".png")
    
    val_path = os.path.join(IMAGE_DIR, val_filename)
    base_path = os.path.join(IMAGE_DIR, base_filename)
    
    if os.path.exists(base_path):
        img_main = Image.open(base_path)
        img_val = Image.open(val_path)
        
       
        new_val_width = int(img_val.width * VAL_SCALE)
        new_val_height = int(img_val.height * VAL_SCALE)
        img_val_resized = img_val.resize((new_val_width, new_val_height), Image.Resampling.LANCZOS)
        
       
        canvas_width = img_main.width
        canvas_height = img_main.height + PADDING_BETWEEN + img_val_resized.height
        
        combined_img = Image.new("RGB", (canvas_width, canvas_height), (255, 255, 255))
        
        combined_img.paste(img_main, (0, 0))
        
        # 6. Căn giữa ảnh validation ở bên dưới
        val_x = (canvas_width - img_val_resized.width) // 2
        val_y = img_main.height + PADDING_BETWEEN
        combined_img.paste(img_val_resized, (val_x, val_y))
        
        # 7. Lưu ảnh
        output_filename = f"{base_filename}"
        combined_img.save(os.path.join(OUTPUT_DIR, output_filename))
        print(f"Đã ghép xong: {output_filename}")
    else:
        print(f"Không tìm thấy file tương ứng cho: {val_filename}")

print("Hoàn tất!")
```
